```python
import pandas as pd
import os
```

```python
data_path = "../data/raw"

csv_files = [f for f in os.listdir(data_path) if f.endswith(".csv")]

print("Total Files:", len(csv_files))
print(csv_files)
```

```python
dataframes = {}

for file in csv_files:

    df = pd.read_csv(os.path.join(data_path, file))

    dataframes[file] = df

    print("\n" + "="*60)
    print("FILE:", file)

    print("\nSHAPE:")
    print(df.shape)

    print("\nDATA TYPES:")
    print(df.dtypes)

    print("\nFIRST 5 ROWS:")
    print(df.head())

    print("\nMISSING VALUES:")
    print(df.isnull().sum())

    print("\nDUPLICATES:")
    print(df.duplicated().sum())
```

```python
summary = []

for file in csv_files:

    df = pd.read_csv(os.path.join(data_path, file))

    summary.append({
        "Dataset": file,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing_Values": df.isnull().sum().sum(),
        "Duplicate_Rows": df.duplicated().sum()
    })

summary_df = pd.DataFrame(summary)

summary_df
```

```python
import pandas as pd
import os

data_path = "../data/raw/"
csv_files = [f for f in os.listdir(data_path) if f.endswith('.csv')]

dataframes = {}
for file in csv_files:
    name = file.replace('.csv', '')
    df = pd.read_csv(os.path.join(data_path, file))
    dataframes[name] = df
    print(f"\n{'='*40}")
    print(f"FILE: {file}")
    print(f"Shape: {df.shape}")
    print(f"Dtypes:\n{df.dtypes}")
    print(f"Head:\n{df.head()}")
    print(f"Nulls:\n{df.isnull().sum()}")
    print(f"Anomalies: {df.duplicated().sum()} duplicate rows")
```
